# Overview of the Dataset

```python
import os
import pandas as pd

data_path = os.path.join(os.getcwd(), '..', '01_Data')
dataset = os.path.join(data_path, '2024.csv')
df = pd.read_csv(dataset)
#df.dropna(inplace=True)
df.head()
```

## High Level Data

```python
print(f"[Overview.1.0] Complete dataset (author x paper): {len(df)}")
print(f"[Overview.1.1] Number of (distinct) authors: {len(set(df['name'].tolist()))}")
print(f"[Overview.1.2] Number of (distinct) publications: {len(set(df['title'].tolist()))}")
print(f"[Overview.1.2] Number of (distinct) years: {len(set(df['year'].tolist()))}")
```

```python
import matplotlib.pyplot as plt

df = pd.read_csv(os.path.join(data_path, 'publications_per_year.csv'))
df = df.sort_values("year")

ax = df.plot.bar(x="year", y="papers", rot=90, figsize=(16, 5), legend=False)
ax.set_xlabel("Year")
ax.set_ylabel("Number of Papers")
ax.set_title("Paper per Year")
plt.tight_layout()
plt.show()
```
